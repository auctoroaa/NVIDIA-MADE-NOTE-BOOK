# NVIDIA API smoke test

Checks configuration, then sends one small chat-completion request to the NVIDIA OpenAI-compatible API. Set `NVIDIA_API_KEY` and `NVIDIA_MODEL` in a local `.env` file first.

```python
import os
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("NVIDIA_API_KEY", "").strip()
MODEL = os.getenv("NVIDIA_MODEL", "").strip()
BASE_URL = os.getenv("NVIDIA_BASE_URL", "https://integrate.api.nvidia.com/v1").strip()

if not API_KEY:
    raise ValueError("NVIDIA_API_KEY is missing. Add it to .env and restart the kernel.")
if not MODEL:
    raise ValueError("NVIDIA_MODEL is missing. Add the exact NVIDIA model identifier to .env.")

print(f"Base URL: {BASE_URL}")
print(f"Model: {MODEL}")
print("API key: configured (value hidden)")
```

```python
from openai import OpenAI

client = OpenAI(base_url=BASE_URL, api_key=API_KEY)
response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": "You are a concise assistant."},
        {"role": "user", "content": "Reply with: NVIDIA API connection works."},
    ],
    max_tokens=64,
    temperature=0.2,
)

print(response.choices[0].message.content)
```

## If it fails

- **401/403:** verify the API key and that the account can access the selected model.
- **404/model error:** check the exact model identifier and endpoint URL in NVIDIA API Catalog.
- **429:** rate or quota limit; wait and retry later.
- **Timeout/network error:** retry and check the service status/network.

Never paste your API key into notebook output or commit it to Git.
